```python
# STEP 0: put the kit at /kaggle/working/film_kit_v3 (auto-finds the uploaded zip or folder)
import os, glob, zipfile, shutil
DEST='/kaggle/working/film_kit_v3'
if not os.path.isdir(DEST):
    folders=[f for f in glob.glob('/kaggle/input/**/film_kit_v3', recursive=True) if os.path.isdir(f)]
    if folders:
        shutil.copytree(folders[0], DEST); print('copied kit from', folders[0])
    else:
        zips=glob.glob('/kaggle/input/**/*film_kit*.zip', recursive=True)+glob.glob('/kaggle/working/*film_kit*.zip', recursive=True)
        assert zips, 'Add THE_ONE_full_film_kit.zip as a Kaggle Dataset (Add Input) first.'
        with zipfile.ZipFile(zips[0]) as z: z.extractall('/kaggle/working')
        print('unzipped', zips[0])
assert os.path.isdir(DEST), DEST+' missing'
print('kit ready ->', DEST)
```
